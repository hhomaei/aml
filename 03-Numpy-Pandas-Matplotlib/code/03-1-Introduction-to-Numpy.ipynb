## Inverse

````python
def inverse(arr):
    det = np.linalg.det(arr)
    rank = np.linalg.matrix_rank(arr)
    if det == 0:
        print(f"Inverse doesn't exist. Determinant is zero. Rank = {rank} for a {arr.shape} matrix")
    else:
        print(f"determinant = {det}, rank = {rank} (full ranked)")
        print(np.linalg.inv(arr))
````

````python
z = np.array([[1, 1, 1], [1, 1, 2], [1, 1, 3]])
inverse(z)
````

````python
x = np.array([[1, 1, 2],
              [3, -3, -1],
              [2, -1, 6]])
inverse(x)
````

**Correlation**

1- Covariance: Shows the dependancy between two variables
$$Cov(x, y) = \frac{1}{n-1} \sum_{i=1}^n\left(x_i-\mu_x\right)\left(y_i-\mu_y\right)$$
Covariance matrix:
$$C=\left[\begin{array}{cc}
Var(x) & Cov(x, y) \\
Cov(x, y) & Var(y)
\end{array}\right]$$

````python
X = np.array([1, 2, 3, 4, 5]) # => mean = 3; var = 1^2 + 2^2 + 0^2 + 1^2 + 2^2 = 10/4 = 2.5
Y = np.array([5, 6, 7, 8, 9]) # => mean = 7; var = 1^2 + 2^2 + 0^2 + 1^2 + 2^2 = 10/4 = 2.5
print (f"Covariance (X, Y) = {np.cov(X, Y)}") # => Cov(X, Y) = (2*2 + 1*1 + 0*0 + 1*1 + 2*2)/4 = 10/4 = 2.5
print (f"var(X) = {np.var(X)}")
````

**Note**: The `var` function with its default parameters calculates $\frac{1}{N} \sum(...)$, but the `cov` function uses $\frac{1}{n-1} \sum(...)$ to compute the variance. If you want to calculate the sample variance ($\frac{1}{n-1} \sum(...)$) as is in the `cov`, you should pass `ddof= 1` argument to the `var` function.

2- Pearson's correlation coefficient
$$\rho_{X, Y}=\frac{Cov(X, Y)}{\sigma_X \sigma_Y}$$
Pearson's correlation coefficient matrix:
$$C=\left[\begin{array}{cc}
1 & \rho_{x, y} \\
\rho_{x, y} & 1
\end{array}\right]$$


````python
print (f"Pearson's correlation coefficient matrix for (X, Y) = {np.corrcoef(X, Y)}")
````

**Matrix Product** <br>
sum of the multiplication of row elements to column element<br>
Example: for $a = [[1, 2], [3, 4]]$, $a \times a = [[1\times 1 + 2\times 3, 1\times 2 + 2\times 4], [3\times 1 + 4\times 3, 3\times 2 + 4\times 4]]$<br>
The general operator is @, however you can also use ```matmul``` function of numpy to get the matrix product of two arrays.

````python
a = np.array([[1, 2], [3, 4]])
a@a # or np.matmul(a, b)
````

Multiple chained multiplication is also available for square matrix. The function ```matrix_power``` raises a square matrix to the (integer) power n. <br>
For positive integers $n$, the power is computed by repeated matrix multiplications. If $n = 0$, the identity matrix of the same shape as M is returned. If $n<0$, the inverse is computed and then raised to the $|n|$.

````python
np.linalg.matrix_power(a, 2)
````

````python
np.linalg.matrix_power(a, 0)
````
